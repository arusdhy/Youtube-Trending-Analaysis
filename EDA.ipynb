```python
import numpy as np, pandas as pd #Imports NumPy and Pandas for numerical operations and data handling
import matplotlib.pyplot as plt, seaborn as sns #Imports Matplotlib and Seaborn for data visualisation
import warnings; warnings.filterwarnings('ignore') #Ignores warning messages

#Displays plots directly in the notebook
%matplotlib inline

import utils #Imports custom utility functions

df=utils.load_data() #Load the dataset
print('shape:',df.shape) #Display the dataset dimensions (rows, columns)
df.head() #Show the first 5 rows of the dataset
```

```python
#Structure & dtypes
print(df.dtypes.value_counts().to_dict()) #Displays the number of columns for each data type
df.info() #Displays dataset information
```

```python
#Missing values
mr = utils.missing_report(df)#Generates a missing values report
print('columns with missing values:', len(mr))#Displays the number of columns with missing values
print(mr.head(10).to_string()) #Displays the top 10 columns with the most missing values
```

```python
#Numeric distributions
#Selects numerical columns for analysis
num = ['views', 'likes', 'dislikes', 'comments']
num = [c for c in num if c in df.columns]

#Creates histograms for each numerical column
fig, ax = plt.subplots(1, len(num), figsize=(4 * len(num), 3))
ax = np.atleast_1d(ax)

#Plots the distribution of each numerical column
for a, c in zip(ax, num):
    a.hist(df[c].dropna(), bins=40, color='#4c72b0')
    a.set_title(c)

#Adjusts layout and display the plots
plt.tight_layout()
plt.show()

df[num].describe().round(2) #Displays summary statistics for the numerical columns
```

```python
#Top categories
vc = utils.top_counts(df['ytvideoid'], 10) #Finds the 10 most common YouTube video IDs

#Creates a bar chart of the most common video IDs
fig, ax = plt.subplots(figsize=(9, 3.5))
vc.plot(kind='bar', ax=ax, color='#55a868')
ax.set_title('Top ytvideoid')

#Formats and display the chart
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

print(vc.to_string()) #Displays the top video IDs and their counts
```
